```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import yfinance as yf
import datetime as dt
```

```python
#IMPORTANDO E FORMATANDO BANCO
end_date = dt.datetime.now()
start_date = end_date - dt.timedelta(days= 7*365)

db = yf.download(["^BVSP"],
                 start= start_date, end= end_date)['Close']
db.columns.name = None

db = db.rename(columns={'^BVSP': 'Close - IBOVESPA'})

db = db.reset_index()
```
